```python
import random

def task2_1(filename, quantity, maximum):
    nums = []
    for i in range(quantity):
        num = random.randint(0, maximum)
        nums.append(num)
        
    with open(filename, 'w') as f:
        for num in nums:
            insert = num + "\n"
            f.write(insert)
    #Closes automatically
    
```

```python
def task2_2(list_of_integers):
    if len(list_of_integers) <= 1:
        return list_of_integers
    
    mid = len(list_of_integers) // 2
    
    left = task2_2(list_of_integers[:mid])
    right = task2_2(list_of_integers[mid:])
    
    return merge(left, right)
    
def merge(left, right):
    l = 0
    r = 0
    results = []
    
    while l < len(left) and r < len(right): #so we stop when either l or r is done checking
        if left[l] > right[r]:
            results.append(right[r])
            r += 1
        else:
            results.append(left[l])
            l += 1
    
    #insert the remaining on either lists in
    results.extend(right[r:])
    results.extend(left[l:])
    
    return results

#main
task2_2([56,25,4,98,0,18,4,5,7,0])
```

```python
def task2_3(filename_in, filename_out):
    data = []
    
    f = open(filename_in, 'r')
    for line in f:
        line = line.strip().split(',')
        data.append(line)
    
    sorted_data = task2_2(data)
    
    wf = open(filename_out, 'w')
    for line in sorted_data:
        wf.write(line)
```

```python
def task3_1(quantity_of_data):
    digit = ""
    end = ""
    for i in quantity_of_data:
        if i.isdigit():
            digit += i
        else:
            end += i
    #print(digit)
    #print(end)
    
    accept = {
        "KB" : 10**3,
        "MB" : 10**6,
        "GB" : 10**9, 
        "TB" : 10**12
    }
    
    if end in accept:
        final = int(digit) * accept[end]
        print(final)
    else:
        print("invalid data")
        
task3_1("8KB")
```

```python
def task3_2(quantity_of_data):
    digit = ""
    end = ""
    for i in quantity_of_data:
        if i.isdigit():
            digit += i
        else:
            end += i
    #print(digit)
    #print(end)
    
    accept = {
        "KB" : 10**3,
        "MB" : 10**6,
        "GB" : 10**9, 
        "TB" : 10**12,
        "KiB" : 2**10,
        "MiB" : 2**20,
        "GiB" : 2**30,
        "TiB" : 2**40,
    }
    
    if end in accept:
        final = int(digit) * accept[end]
        return final
    else:
        return "invalid data"


print(task3_2("2MiB"))
```

```python
def task3_3(quantity_of_data, target_unit):
    accept = {
        "KB" : 10**3,
        "MB" : 10**6,
        "GB" : 10**9, 
        "TB" : 10**12,
        "KiB" : 2**10,
        "MiB" : 2**20,
        "GiB" : 2**30,
        "TiB" : 2**40,
    }
    
    if target_unit not in accept:
        return "invalid data"
    
    result = task3_2(quantity_of_data)
    if type(result) == int:
        result = result / accept[target_unit]
        return result
    else:
        return "invalid data"

task3_3("512MiB", "GiB")
```
