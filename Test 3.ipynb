```python
#Task 1 corrections
import pymongo
import json

#task 1.1
client = pymongo.MongoClient("127.0.0.1", 27017)
db = client["ParcelDB"]
parcels = db["Parcels"]
```

```python
#Task 1.2
def loadparcels():
    """Read parcel records from Task1_2.json and insert them into MongoDB""" 
    #This^ is just like a comment, but for professionals
    
    
    with open("TASK1_2.JSON", 'r') as file:
        parcel_data = json.load(file)
    #File close automatically
        
    #Clear existing documments so repeated testing does not create duplicates
    parcels.delete_many({})
    
    if len(parcel_data) > 0:
        parcels.insert_many(parcel_data)
        print("Parcels records loaded successfully")
        
    else:
        print("No parcel records found in JSON file")
```

```python
#Task 1.3
def displayParcel(parcelID):
    """Search for and display one parcel based on parcelID"""
    
    parcel = parcels.find_one({"parcelID":parcelID})
    
    if parcel != None:
        print("---Parcel Found---")
        print("Parcel ID: ", parcel["parcelID"])
        print("Recipient: ", parcel["recipient"])
        print("Priority: ", parcel["priority"])
        print("Status: ", parcel["status"])
    else:
        print("Parcel Not Found")
```

```python
#Task 1.4
def priorityParcels(priorityType):
    results = parcels.find({'priority': priorityType})
    count = 0
    
    print(f"{'Parcel ID':<12} {'Recipient':<20}")
    
    for parcel in results:
        print(f"{parcel['parcelID']:<12} {parcel['recipient']:<20}")
        count += 1
        
    print("Total no. of parcels: ", count)
```
