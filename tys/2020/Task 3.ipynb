```python
#Task 3.1
def task3_1(quantity_of_data):
    digit = ""
    end = ""
    for i in quantity_of_data:
        if i.isdigit():
            digit += i
        else:
            end += i
    #print(digit)
    #print(end)
    
    accept = {
        "KB" : 10**3,
        "MB" : 10**6,
        "GB" : 10**9, 
        "TB" : 10**12
    }
    
    if end in accept:
        final = int(digit) * accept[end]
        print(final)
    else:
        print("invalid data")
    
#Main
task3_1("8KB")
```

```python
#Task 3.2
def task3_2(quantity_of_data):
    digit = ""
    end = ""
    for i in quantity_of_data:
        if i.isdigit():
            digit += i
        else:
            end += i
    #print(digit)
    #print(end)
    
    accept = {
        "KB" : 10**3,
        "MB" : 10**6,
        "GB" : 10**9, 
        "TB" : 10**12,
        "KiB" : 2**10,
        "MiB" : 2**20,
        "GiB" : 2**30,
        "TiB" : 2**40,
    }
    
    if end in accept:
        final = int(digit) * accept[end]
        return final
    else:
        return "invalid data"

#Main
print(task3_2("2MiB"))
```

```python
#Task 3.3
def task3_3(quantity_of_data, target_unit):
    accept = {
        "KB" : 10**3,
        "MB" : 10**6,
        "GB" : 10**9, 
        "TB" : 10**12,
        "KiB" : 2**10,
        "MiB" : 2**20,
        "GiB" : 2**30,
        "TiB" : 2**40,
    }
    
    if target_unit not in accept:
        return "invalid data"
    
    result = task3_2(quantity_of_data)
    if type(result) == int:
        result = result / accept[target_unit]
        return result
    else:
        return "invalid data"

#Main
task3_3("512MiB", "GiB")
```
