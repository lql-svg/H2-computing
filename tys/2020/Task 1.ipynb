```python
#Task 1.1
def task1_1(string_value):
    h = 0
    for i in range(len(string_value)):
        val = 33 * (ord(string_value[i]))
        h = (h + val) % 1024
    return h
                    
task1_1("Hello") 
```

```python
#Task 1.2
def task1_2(seed, string_value):
    new_str = seed + string_value
    return task1_1(new_str)
```
