```python
#Task 2.1
import random

def task2_1(filename, quantity, maximum):
    nums = []
    for i in range(quantity):
        num = random.randint(0, maximum)
        nums.append(num)
        
    with open(filename, 'w') as f:
        for num in nums:
            insert = str(num) + "\n"
            f.write(insert)
    #Closes automatically
    
#main
task2_1("randomnumbers.txt", 1000, 5000)
```

```python
#Task 2.2
def task2_2(list_of_integers):
    if len(list_of_integers) <= 1:
        return list_of_integers
    
    mid = len(list_of_integers) // 2
    
    left = task2_2(list_of_integers[:mid])
    right = task2_2(list_of_integers[mid:])
    
    return merge(left, right)
    
def merge(left, right):
    l = 0
    r = 0
    results = []
    
    while l < len(left) and r < len(right): #so we stop when either l or r is done checking
        if left[l] > right[r]:
            results.append(right[r])
            r += 1
        else:
            results.append(left[l])
            l += 1
    
    #insert the remaining on either lists in
    results.extend(right[r:])
    results.extend(left[l:])
    
    return results

#main
task2_2([56,25,4,98,0,18,4,5,7,0])
```

```python
#Task 2.3
def task2_3(filename_in, filename_out):
    data = []
    
    f = open(filename_in, 'r')
    for line in f:
        line = line.strip()
        data.append(int(line))
    
    sorted_data = task2_2(data)
    #print(sorted_data)
    
    wf = open(filename_out, 'w')
    for line in sorted_data:
        num = str(line) + '\n'
        wf.write(num)
        
#main
task2_3("randomnumbers.txt", "sortednumbers.txt")
```
