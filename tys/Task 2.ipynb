```python
#Task 2.1
class Node:
    def __init__(self):
        self.data_value = -1
        self.left_node = -1
        self.right_node = -1
        
    def set_data(self, d):
        self.data_value = d
        
    def set_left(self, l):
        self.left_node = l
        
    def set_right(self, r):
        self.right_node = r
        
    def get_data(self):
        return self.data_value
    
    def get_left(self):
        return self.left_node
    
    def get_right(self):
        return self.right_node

    
```

```python
#Task 2.2
def calculate_nodes(levels):
    total = 2 ** (levels) - 1
    print(f"A {levels} level number tree has {total} nodes")
    return total


#task 2.3
calculate_nodes(1)
calculate_nodes(3)
calculate_nodes(5)
calculate_nodes(10)
```

```python
#task 2.4
def create_empty_tree(list_nodes, levels):
    nodes = calculate_nodes(levels)
    
    for i in range(nodes):
        node = Node()
        list_nodes.append(node)
        
    return list_nodes
    
tree = create_empty_tree([], 3)
```

```python
#task 2.5
def assign_leaf_nodes(list_nodes, levels):
    leaf_nodes = 2 ** (levels - 1)
    
    for i in range(leaf_nodes):
        list_nodes[i].set_data(i + 1)
        
    return list_nodes

tree = assign_leaf_nodes(tree, 3)
```

```python
#task 2.6
def create_number_tree(list_nodes, levels): #List already has leaf nodes
    num_leaf = 2 ** (levels - 1)
    num_nodes = len(list_nodes)
    index = num_leaf
    
    for i in range(0, num_nodes - 1, 2):
        #Get summation first
        #print("First at: ", list_nodes[i].get_data())
        #print("Second: ", list_nodes[i + 1].get_data())
        summation = list_nodes[i].get_data() + list_nodes[i + 1].get_data()
        #print(summation)
        
        #Insert into list
        list_nodes[index].set_data(summation)
        list_nodes[index].set_left(i)
        list_nodes[index].set_right(i + 1)
        index += 1
        
    return list_nodes
     
    
#main
tree = create_number_tree(tree, 3)
```

```python
#Task 2.7
tree = create_empty_tree([], 3)
tree = assign_leaf_nodes(tree, 3)
tree = create_number_tree(tree, 3) #Added everything


print(f"{'Left ptr':^10} | {'Data':^10} | {'Right ptr':^10}")
for i in range(len(tree)):
    print(f"{tree[i].get_left():^10} | {tree[i].get_data():^10} | {tree[i].get_right():^10}")
    
```
