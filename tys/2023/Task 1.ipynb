```python
#Task 1.1
import sqlite3

conn = sqlite3.connect("TESTDATABASE.db")

conn.execute("DROP TABLE IF EXISTS Student_Test")
conn.execute("DROP TABLE IF EXISTS Student")
conn.execute("DROP TABLE IF EXISTS Test")
conn.execute("DROP TABLE IF EXISTS Class")

#create
student = """CREATE TABLE Student(
            StudentID INTEGER PRIMARY KEY,
            GivenName TEXT, 
            FamilyName TEXT,
            ClassName TEXT,
            FOREIGN KEY (ClassName) REFERENCES Class)"""

test = """CREATE TABLE Test(
            TestID TEXT PRIMARY KEY,
            MaxMarks INTEGER)"""

Class = """CREATE TABLE Class(
            ClassName TEXT PRIMARY KEY,
            YearGroup INTEGER)"""

st = """CREATE TABLE Student_Test(
        StudentTestID INTEGER PRIMARY KEY,
        StudentID INTEGER,
        TestID TEXT,
        Mark INTEGER,
        DateTest TEXT,
        FOREIGN KEY (StudentID) REFERENCES Student,
        FOREIGN KEY (TestID) REFERENCES Test)""" 

#Execute create
conn.execute(test)
conn.execute(Class)
conn.execute(student)
conn.execute(st)

conn.commit()
conn.close()
```
