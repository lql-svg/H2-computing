```python
#Task 2.1
class QueueItem:
    def __init__(self, ID, qty):
        self.ID = ID
        self.quantity = qty
    
    def getID(self):
        return self.ID
    
    def getQuantity(self):
        return self.quantity
```

```python
#Task 2.2
class Queue: #CIRCUlAR
    def __init__(self):
        self.headPointer = 0
        self.tailPointer = 0
        self.numberItems = 0
        
        self.queueList = [None] * 4
        
        
    #Task 2.3
    def isEmpty(self):
        if self.numberItems == 0:
            return True
        return False
    
    def isFull(self):
        if self.numberItems == 4:
            return True
        return False
        
        
    def enqueue(self, queueItem): #Queue item has the id and qty
        if self.isFull():
            return False
            
        #else, shift head
        self.queueList[self.tailPointer] = queueItem
        self.tailPointer = (self.tailPointer + 1) % 4
        self.numberItems += 1
        return True

    #Task 2.4
    def dequeue(self):
        if self.isEmpty():
            return False
        
        else:
            first = self.queueList[self.headPointer].quantity
            self.headPointer = (self.headPointer + 1) % 4
            self.numberItems -= 1
            return first
        
    #Task 2.5
    def printQueue(self):
        index = self.headPointer
        print(f"{'ID':^5} | {'Quantity':^10}")

        for i in range(self.numberItems):
            item = self.queueList[index]
            print(f"{item.getID():^5} | {item.getQuantity():^10}")
            index = (index + 1) % 4
```

```python
#Task 2.3
q = Queue()

for i in range(5):
    ID = int(input("Input ID: "))
    qty = int(input("Input Quantity: "))
    
    item = QueueItem(ID, qty)
    
    sucess = q.enqueue(item)
    
    if sucess == True:
        print("Success")
    else:
        print("Failure")

```

```python
#2.4
q = Queue()
for i in range(5):
    ID = i + 1
    qty = i + 1
    item = QueueItem(ID, qty)
    q.enqueue(item)

for i in range(5):
    output = q.dequeue()
    print(output)
```
