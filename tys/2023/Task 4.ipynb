```python
#Task 4.1
def read(FileName):
    data = []
    
    file = open(FileName, 'r')
    for line in file:
        line = line.strip()
        data.append(line)
    
    file.close()
    
    return data

def compress(data):
    output = ""
    current = data[0]
    amt = 1
    
    for i in range(1, len(data)): #Get all in data
        if data[i] != current: #If next one is no longer like current
            output += current #Input current first
            output += str(amt) #Input amt
            amt = 1 #Reset amt
            
            current = data[i] #Update current
            
        else: #if all are same
            amt += 1
            
    #Last set
    output += current #Input current first
    output += str(amt) #Input amt
    return output
            

def write_compressed(compressed_data):
    wf = open("compressedimage.txt", 'w') #Write file
    wf.write(compressed_data)
    wf.close()
    
#main
data = read("imagefile.txt")
compressed = compress(data)
write_compressed(compressed)
```

```python
#task 4.2
def read(FileName):
    data = []
    
    file = open(FileName, 'r')
    for line in file:
        data.append(line)
    file.close()
    
    data = data[0] #Get a string instead
    return data

def decompress(data):
    decompressed = []
    for item in range(3, len(data), 4):
        code = data[item - 3:item] #Code
        count = data[item] #Count of number of times we got the data
        
        for i in range(int(count)):
            decompressed.append(code)
            
    return decompressed

def write_decompressed(decompressed_data):
    wf = open("decompressedimage.txt", 'w') #Write file
    
    for line in decompressed_data:
        wf.write(line + '\n')
    wf.close()


#main testing
data = read("compressedimage.txt")
#print(data)
decompressed = decompress(data)
#print(decompressed)

#Menu
file_name = input("File name: ")
data = read(file_name)

option = input("Compression / Decompression: ")

if option.lower() == "compression":
    compressed = compress(data)
    print("compressed")

    
elif option.lower() == "decompression":
    decompressed = decompress(data)
    write_decompressed(decompressed)
    print("decompressed")

else: 
    print("Invalid option")
```
