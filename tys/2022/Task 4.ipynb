```python
#task 4.1
import sqlite3

conn = sqlite3.connect("2022_LIBRARY.db")

#Drop
conn.execute("DROP TABLE IF EXISTS Loan")
conn.execute("DROP TABLE IF EXISTS Book")
conn.execute("DROP TABLE IF EXISTS Member")


book_sql = """CREATE TABLE Book(
                BookID INTEGER PRIMARY KEY NOT NULL,
                Title TEXT NOT NULL,
                Genre TEXT NOT NULL)
            """

member_sql = """CREATE TABLE Member(
                MemberNumber INTEGER PRIMARY KEY NOT NULL,
                FamilyName TEXT NOT NULL,
                GivenName TEXT NOT NULL
                )"""

loan_sql = """CREATE TABLE Loan(
                LoanID INTEGER PRIMARY KEY NOT NULL,
                MemberNumber INTEGER NOT NULL,
                BookID INTEGER NOT NULL, 
                DateLoaned DATE NOT NULL,
                Returned TEXT NOT NULL,
                FOREIGN KEY (MemberNumber) REFERENCES Member(MemberNumber),
                FOREIGN KEY (BookID) REFERENCES Book(BookID)
                )"""

conn.execute(book_sql)
conn.execute(member_sql)
conn.execute(loan_sql)

conn.commit()
conn.close()
```

```python
#Task 4.2

def read(fileName):
    data = []
    
    file = open(fileName, 'r')
    for line in file:
        line = line.strip().split(",")
        data.append(line)
        
    file.close()
    return data

#connection
conn = sqlite3.connect("2022_LIBRARY.db")

books = read("BOOK.txt")
member = read("MEMBER.txt")
loan = read("LOAN.txt")

for i in books:
    ID = i[0]
    name = i[1]
    genre = i[2]
    
    conn.execute("INSERT INTO Book VALUES (?, ?, ?)", (ID, name, genre))

for i in member:
    ID = i[0]
    f_name = i[1] #Family name
    g_name = i[2] #Given name
    #print(i)
    
    conn.execute("INSERT INTO Member VALUES (?, ?, ?)", (ID, f_name, g_name))
    
for i in loan:
    ID = i[0]
    memID = i[1]
    bookID = i[2]
    date = i[3]
    returned = i[4]
    
    
    conn.execute("INSERT INTO Loan VALUES (?, ?, ?, ?, ?)", 
                 (ID, memID, bookID, date, returned))


conn.commit()
conn.close()
```
