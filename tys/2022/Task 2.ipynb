```python
#Task 2.2
def bubble_sort(ar):
    sort = 0
    
    while True: #Keep passing
        swap = False
        
        for i in range(len(ar) - 1 - sort): #Each pass
            if ar[i] > ar[i + 1]:
                ar[i], ar[i + 1] = ar[i + 1], ar[i]
                swap = True
                
        sort += 1
        
        if swap == False: #To be able to break early if possible
            break

    return ar

#main
data = task2_1()
print(bubble_sort(data))
```

```python
#Task 2.3
def merge_sort(ar):
    if len(ar) <= 1:
        return ar
    
    mid = len(ar) // 2
    
    left = merge_sort(ar[:mid])
    right = merge_sort(ar[mid:])
    
    return merge(left, right)

def merge(left, right):
    l = 0
    r = 0
    results = []
    
    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            results.append(left[l])
            l += 1
        else:
            results.append(right[r])
            r += 1
        
    results.extend(left[l:])
    results.extend(right[r:])
    
    return results

#main
data = task2_1()
while True:
    print("1. Bubble Sort")
    print("2. Merge Sort")
    user_input = input("Select an option: ")
    
    if user_input == "1":
        output = bubble_sort(data)
        print(output)
        break
        
    elif user_input == "2":
        output = merge_sort(data)
        print(output)
        break
    
```

```python
#Task 2.4
def binary_search(ar, low, high, target):
    if low > high:
        return -1 #Not found
    
    mid = (low + high) // 2

    if ar[mid] == target:
        return mid
    
    elif ar[mid] > target: #go left
        return binary_search(ar, low, mid - 1, target)
    elif ar[mid] < target: #go right
        return binary_search(ar, mid + 1, high, target)
        

#main
data = task2_1() #Call
sort = merge_sort(data) #sort data first for binary search 
#print(sort)

user_input = int(input("Input an integer: ")) #Get user input
output = binary_search(sort, 0, 99, user_input) #Data is global of size 100

if output == -1:
    print("Not found")
else:
    print("Found at index: ",output)
```
