```python
#Task 3.1
class Tree:
    def __init__(self, data):
        self.left_pointer = None
        self.right_pointer = None
        self.data = data
        
        
    def insert(self, new_data):
        #Traverse left
        if new_data < self.data: #Less than data
            if self.left_pointer == None: #If left is none, insert
                self.left_pointer = Tree(new_data)
                
            else: #Continue traversing, recurse left
                self.left_pointer.insert(new_data)
                
        else: #More than data
            if self.right_pointer == None: #If left is none, insert
                self.right_pointer = Tree(new_data)
                
            else: #Continue traversing, recurse left
                self.right_pointer.insert(new_data)

    
    def inorder(self):
        if self.left_pointer != None:
            self.left_pointer.inorder() #Keep traversing down left first
        
        print(self.data) #Print the data
        
        if self.right_pointer != None:
            self.right_pointer.inorder() #Keep traversing down right
    
    def postorder(self):
        if self.left_pointer != None:
            self.left_pointer.postorder() #Keep traversing down left first
            
        if self.right_pointer != None:
            self.right_pointer.postorder() #Keep traversing down right
            
        print(self.data) #Print the data


#Main
import random


nums = []
while len(nums) < 10: #Go until length of nums is 10
    number = random.randint(0, 999)
    
    #Checking for unique number
    if number not in nums:
        nums.append(number) #put into array
        
#print(nums)

#Tree
tree = Tree(nums[0])
for i in nums[1:]:
    tree.insert(i)
    
tree.inorder()
print()
tree.postorder()
```
