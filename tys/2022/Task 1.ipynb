```python
#Task 1.1
def encrypt(char):
    #First check
    if char == " ": #Force spaces to "!"
        num = 33
        return num

    #Then check other cases
    if char.isalpha() == False or len(char) != 1: #eliminate non-string
        return -1
    
    num = ord(char)
    num += 10
    
    #Upper: 65-90
    #Lower: 97-122
    if char.isupper():
        if num > 90:
            num -= 26
    
    elif char.islower():
        if num > 122:
            num -= 26

    #print(chr(num))
    return num
```

```python
#Task 1.2
print(encrypt("A"))
print(encrypt("a"))
print(encrypt("#"))
print(encrypt(" "))
```

```python
#Task 1.3

#read
data = []
file = open("DATATOENCRYPT.txt", 'r')

for line in file:
    line = line.strip()
    data.append(line)
    
file.close()

#encryption
msg = data[0]
#print(msg)
encrypted = ""

for i in msg:
    #print(i)
    num = encrypt(i)
    
    if num != -1: #so if encryption is valid
        encrypted += chr(num) 
        #I'm assuming we return in text form and not a line of integer
        
#print(encrypted)

#Write
wf = open("ENCRYPTEDMESSAGE.txt", 'w')
wf.write(encrypted)
wf.close()
```
