```python
def task1_1(input_value):
    input_value = str(input_value)
    
    #Validation
    if input_value.isdigit() == False:
        #print("not digits")
        return -1
    if len(input_value) != 14:
        #print("Not 14 digits long")
        return -1
    
    #Double + add in a string
    total_str = ""
    for i in range(14):
        if i % 2 != 0: #odd a bit wrong
            value = str(int(input_value[i]) * 2)
        else:
            value = str(int(input_value[i]))
            
        total_str += value
    
    #Add all digits
    total = 0
    for num in total_str:
        total += int(num)
    
    #check digit
    check = (10 - (total % 10)) % 10 #second %10 to ensure if check digit is not 10
    return check

```

```python
#1.2
print(task1_1("14576567652234") == 9)
print(task1_1(14916567654934) == 1)
print(task1_1("145r76567654934") == -1)
print(task1_1("146567654934") == -1)
```

```python
def task1_3(input_value):
    input_value = str(input_value)
    
    #Validation
    if input_value.isdigit() == False:
        #print("not digits")
        return False
    if len(input_value) != 15:
        #print("Not 15 digits long")
        return False
    
    last = input_value[-1] #Get last digit    
    value = input_value[:-1] #Everything but the last digit
    
    check = task1_1(value)
    
    if int(check) == int(last):
        return True
    else:
        #print("wrong check value")
        return False

    
#Main
print(task1_3("145765676522349"))
print(task1_3(149165676549341))
print(not task1_3("145r76567654934"))
print(not task1_3("146567654934"))
```
