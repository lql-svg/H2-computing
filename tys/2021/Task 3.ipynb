```python
#Task 3.1
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
    

class LinkedList:
    def __init__(self):
        self.root = None
    
    def insert(self, integer_value):
        node = Node(integer_value)
        
        if self.root == None:
            self.root = node
        
        else:
            node.next = self.root #node points to root
            self.root = node #Root goes to node
            
    def delete(self, integer_value):
        if self.root == None:
            print("Nothing in list")
            return None
        
        if self.root.data == integer_value:
            self.root = self.root.next
            return 
            
        current = self.root
        prev = None

        while current != None:
            if current.data == integer_value:
                prev.next = current.next
                current.next = None
                return

            else:
                prev = current
                current = current.next
        
        
    def search(self, integer_value):
        if self.root == None:
            print("Nothing in list")
            return False
        
        elif self.root.next == None:
            if self.root.data != integer_value:
                return False
            return True
            
        else:
            current = self.root
            
            while current != None:
                if current.data == integer_value:
                    return True
                else:
                    current = current.next
                
            return False

    def count(self):
        current = self.root
        count = 0

        while current is not None:
            count += 1
            current = current.next

        return count
        
    def to_String(self):
        data = []
        current = self.root

        while current != None:
            data.append(current.data)
            current = current.next

        return data


l = LinkedList()
l.insert(5)
l.insert(7)
l.insert(1)
l.insert(3)
l.insert(2)
l.insert(8)
l.insert(4)
l.insert(9)
l.insert(6)
l.insert(0)

l.to_String()
```

```python
#Task 3.2
class SortedLinkedList(LinkedList):
    def insert(self, data):
        node = Node(data)
        
        if self.root == None: #Nothing
            self.root = node
        
        elif self.root.data >= data: #Insert at head
            node.next = self.root 
            self.root = node
        
        else: #insert in the body
            current = self.root
            prev = None
            
            while current != None and current.data < data:
                prev = current
                current = current.next

            prev.next = node
            node.next = current
            
#main
sl = SortedLinkedList()
sl.insert(5)
sl.insert(7)
sl.insert(1)
sl.insert(3)
sl.insert(2)
sl.insert(8)
sl.insert(4)
sl.insert(9)
sl.insert(6)
sl.insert(0)

print(sl.to_String())
print(sl.search(94))
```

```python
#Task 3.3
class Queue(LinkedList):
    def enqueue(self, integer_value):
        node = Node(integer_value)
        
        if self.root == None: #If nothing in
            self.root = node
        
        else: #Insert at the end
            current = self.root
            
            while current.next != None: #Stop at the very last item
                current = current.next
            
            current.next = node #Pointer
    
    def dequeue(self):
        if self.root == None:
            return None
        
        else:
            data = self.root.data
            self.root = self.root.next
            
            return data
        
        
#main
q = Queue()
q.insert(5)
q.insert(7)
q.insert(1)
q.insert(3)
q.insert(2)
q.insert(8)
q.insert(4)
q.insert(9)
q.insert(6)
q.insert(0)

for i in range(5):
    print(q.dequeue())
```
