```python
#Task 2.1
def task2_1(filename):
    data = []
    
    file = open(filename, 'r')
    for line in file:
        line = int(line.strip())
        data.append(line)
    
    file.close()
    return data
    
result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
#Task 2.2
def task2_2(list_of_integers):
    for i in range(1, len(list_of_integers)):
        current = list_of_integers[i]
        pos = i - 1
        
        while pos >= 0 and list_of_integers[pos] > current:
            list_of_integers[pos + 1] = list_of_integers[pos]
            pos -= 1
        
        list_of_integers[pos + 1] = current
        
    return list_of_integers
        
task2_2(result)
```

```python
#Task 2.3
def task2_3(list_of_integers):
    low = 0
    high = len(list_of_integers) - 1
    return quicksort(list_of_integers, low, high)

def partition(list_of_integers, low, high):
    pivot = list_of_integers[high]
    ltp = low - 1 #Lower than pivot
    index = low
    
    while index < high:
        if list_of_integers[index] < pivot:
            ltp += 1
            list_of_integers[index], list_of_integers[ltp] = list_of_integers[ltp], list_of_integers[index]
            
        index += 1
    
    #swap to put pivot in
    ltp += 1
    list_of_integers[high], list_of_integers[ltp] = list_of_integers[ltp], list_of_integers[high]
    
    return ltp
    
def quicksort(list_of_integers, low, high):
    if low < high:
        pi = partition(list_of_integers, low, high)
        quicksort(list_of_integers, low, pi - 1)
        quicksort(list_of_integers, pi + 1, high)
        
    return list_of_integers


#main
print(task2_3(task2_1('TEN.txt')))
```

```python
#Task 2.4
import timeit

for filename in ["TEN.txt", "HUNDRED.txt", "THOUSAND.txt"]:
    numbers = task2_1(filename)
    insertion_time = timeit.timeit(lambda: task2_2(numbers), number = 1)
    quicksort_time = timeit.timeit(lambda: task2_3(numbers), number = 1)
    print(filename)
    print("insertion sort:", insertion_time, "seconds")
    print("quick sort:", quicksort_time, "seconds")
```
