```python
def task1_1(input_value):
    input_value = str(input_value)
    
    #Validation
    if input_value.isdigit() == False:
        print("not digits")
        return -1
    if len(input_value) != 14:
        print("Not 14 digits long")
        return -1
    
    #Double + add in a string
    total_str = ""
    for i in range(14):
        if i % 2 != 0: #odd a bit wrong
            value = str(int(input_value[i]) * 2)
        else:
            value = str(int(input_value[i]))
            
        total_str += value
    
    #Add all digits
    total = 0
    for num in total_str:
        total += int(num)
    
    #check digit
    check = (10 - (total % 10)) % 10 #second %10 to ensure if check digit is not 10
    return check

#1.2
print(task1_1("14576567652234"))
print(task1_1(14916567654934))
print(task1_1("145r76567654934"))
print(task1_1("146567654934"))
```

```python
def task1_3(input_value):
    input_value = str(input_value)
    
    #Validation
    if input_value.isdigit() == False:
        print("not digits")
        return False
    if len(input_value) != 15:
        print("Not 15 digits long")
        return False
    
    last_str = len(input_value) - 2
    last = input_value[:last_str:-1] #Get last digit
    
    value = input_value[:(len(input_value) - 1)]
    
    check = task1_1(value)
    
    if int(check) == int(last):
        return True
    else:
        print("wrong check value")
        return False

    
task1_3(145765692549346)
```

```python
#Task 2.1
def task2_1(filename):
    data = []
    
    file = open(filename, 'r')
    for line in file:
        line = int(line.strip())
        data.append(line)
    
    file.close()
    return data
    
result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
#Task 2.2
def task2_2(list_of_integers):
    for i in range(1, len(list_of_integers)):
        current = list_of_integers[i]
        pos = i - 1
        
        while pos >= 0 and list_of_integers[pos] > current:
            list_of_integers[pos + 1] = list_of_integers[pos]
            pos -= 1
        
        list_of_integers[pos + 1] = current
        
    return list_of_integers
        
task2_2(result)
```

```python
#Task 2.3
def task2_3(list_of_integers):
    low = 0
    high = len(list_of_integers) - 1
    return quicksort(list_of_integers, low, high)

def partition(list_of_integers, low, high):
    pivot = list_of_integers[high]
    ltp = low - 1 #Lower than pivot
    index = low
    
    while index < high:
        if list_of_integers[index] < pivot:
            ltp += 1
            list_of_integers[index], list_of_integers[ltp] = list_of_integers[ltp], list_of_integers[index]
            
        index += 1
    
    #swap to put pivot in
    ltp += 1
    list_of_integers[high], list_of_integers[ltp] = list_of_integers[ltp], list_of_integers[high]
    
    return ltp
    
def quicksort(list_of_integers, low, high):
    if low < high:
        pi = partition(list_of_integers, low, high)
        quicksort(list_of_integers, low, pi - 1)
        quicksort(list_of_integers, pi + 1, high)
        
    return list_of_integers


#main
print(task2_3(task2_1('TEN.txt')))
```

```python
#Task 3.1
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
    

class LinkedList:
    def __init__(self):
        self.root = None
    
    def insert(self, integer_value):
        node = Node(integer_value)
        
        if self.root == None:
            self.root = node
        
        else:
            node.next = self.root #node points to root
            self.root = node #Root goes to node
            
    def delete(self, integer_value):
        if self.root == None:
            print("Nothing in list")
            return None
        
        if self.root.data == integer_value:
            self.root = self.root.next
            return 
            
        current = self.root
        prev = None

        while current != None:
            if current.data == integer_value:
                prev.next = current.next
                current.next = None
                return

            else:
                prev = current
                current = current.next
        
        
    def search(self, integer_value):
        if self.root == None:
            print("Nothing in list")
            return False
        
        elif self.root.next == None:
            if self.root.data != integer_value:
                return False
            return True
            
        else:
            current = self.root
            
            while current != None:
                if current.data == integer_value:
                    return True
                else:
                    current = current.next
                
            return False

    def count(self):
        current = self.root
        count = 0

        while current is not None:
            count += 1
            current = current.next

        return count
        
    def to_String(self):
        data = []
        current = self.root

        while current != None:
            data.append(current.data)
            current = current.next

        return data


l = LinkedList()
l.insert(5)
l.insert(7)
l.insert(1)
l.insert(3)
l.insert(2)
l.insert(8)
l.insert(4)
l.insert(9)
l.insert(6)
l.insert(0)

l.to_String()
```

```python
#Task 3.2
class SortedLinkedList(LinkedList):
    def insert(self, data):
        node = Node(data)
        
        if self.root == None: #Nothing
            self.root = node
        
        elif self.root.data >= data: #Insert at head
            node.next = self.root 
            self.root = node
        
        else: #insert in the body
            current = self.root
            prev = None
            
            while current != None and current.data < data:
                prev = current
                current = current.next

            prev.next = node
            node.next = current
            
#main
sl = SortedLinkedList()
sl.insert(5)
sl.insert(7)
sl.insert(1)
sl.insert(3)
sl.insert(2)
sl.insert(8)
sl.insert(4)
sl.insert(9)
sl.insert(6)
sl.insert(0)

sl.to_String()
```

```python
#Task 3.3
class Queue(LinkedList):
    def enqueue(self, integer_value):
        node = Node(integer_value)
        
        if self.root == None: #If nothing in
            self.root = node
        
        else: #Insert at the end
            current = self.root
            
            while current.next != None: #Stop at the very last item
                current = current.next
            
            current.next = node #Pointer
    
    def dequeue(self):
        if self.root == None:
            return None
        
        else:
            data = self.root.data
            self.root = self.root.next
            
            return data
```
